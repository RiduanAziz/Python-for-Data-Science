# Dictionary Operations

- Membership
- Iteration

```python
d = {"name": "Riduan", "age":27, "gender": "male"}
```

```python
"name" not in d
```

```python
for i in d:
    print(i,d[i])
```

```python
for i in d:
    print(f"{i}:{d[i]}")
```

## Dictionary Methods/Functions

```python
d = {"name": "Riduan", "age":27, "gender": "male"}
#common functions
print(len(d))
print(sorted(d,reverse=False))
print(max(d))
print(min(d))
```

## items/keys/values

```python
print(d.items())
```

```python
print(d.keys())
```

```python
print(d.values())
```

## update

```python
d1 = {1:2,3:4,4:5}
d2 = {4:7,6:8}

d1.update(d2)
print(d1)
```
