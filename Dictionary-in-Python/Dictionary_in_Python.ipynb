# Dictionary

Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

In some languages it is known as map or assosiative arrays.

dict = { 'name' : 'Riduan' , 'age' : 27 , 'gender' : 'male' }

### Characterstics:

- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items

```python
# empty dictionary
d = {}
print(d)
print(type(d))
```

```python
# 1D dictionary
d1 = {"name": "Riduan", "age":27, "Gender":"male"}
print(d1)
```

```python
# with mixed keys
d2 = {(1,2,3):33, "name": "Riduan"}
print(d2)
```

```python
# 2D dictionary
d3 = {
    "name": "Riduan",
    "college_name": "ABC",
    "semester": 5,
    "subjects": {"DSA":80, "Math":50, "English": 70}

}

print(d3)
```

```python
# using sequence and dict function
d4 = dict([("name", "Riduan"), ("age", 27)])
print(d4)
```

```python
# duplicate keys
d5 = {"name": "Riduan", "name": "Alex", "age":27, "Gender":"male"}
print(d5)
```

## Accessing items

```python
d1[0]
```

```python
d1["name"]
```

```python
d3["subjects"]["Math"]
```

## Adding key-value pair

```python
d3["subjects"]["python"] = 80
d3
```

## Editing key-value pair

```python
d1['name'] = "Alex"
d1
```

## Remove key-value pair

```python
# pop
d1.pop("age")
print(d1)
```

```python
# popiteam

d1.popitem()
print(d1)
```

```python
# del
d1 = {"name": "Riduan", "age":27, "Gender":"male"}
del d1["name"]
print(d1)
```

```python
#clear

d1 = {"name": "Riduan", "age":27, "Gender":"male"}
d1.clear()
print(d1)
```
