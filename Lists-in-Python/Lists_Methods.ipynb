# Lists Methods/Functions

## Lists is Mutable, That means you can change the list

```python
# append

L = [1,2,3,4]
print(L)
```

```python
L.append(5)
print(L)
```

```python
s = "Riduan"
s[0] = "A"
```

```python
# Edit
L = [1,2,3,4]

L[0] = 100

print(L)
```

```python
# Editing with Slicing

L = [1,2,3,4]

L[1:4] = [200,300,400]

print(L)
```

```python
L = [1,2,3,4]

L.append([6,7,8])
print(L)
```

```python
L.append(True)
```

## Extend

```python
L = [1,2,3,4]

a = [10,11,12]

L.extend(a)

print(L)
```

## Insert

```python
L = [1,2,3,4]

L.insert(1, 200)

print(L)
```

## Deleting items from a List

```python
L = [1,2,3,4]

del L[0]

print(L)
```

```python
L = [1,2,3,4]

del L[1:3]

print(L)
```

## Remove

```python
L = [1,2,3,4]

L.remove(4)

print(L)
```

## Pop

```python
L = [1,2,3,4]

L.pop()
print(L)
```

## Clear

```python
L = [1,2,3,4]

L.clear()

print(L)
```

## len/min/max/sorted

```python
L = [1,2,3,4]

print(len(L))
print(min(L))
print(max(L))
print(sorted(L,reverse=True))
```

## Count 

```python
L = [1,2,3,4,4,4,5,5]

L.count(4)
```

## Index

```python
L = [1,2,3,4,4,4,5,5]

L.index(2)
```

## Reverse

```python
L = [1,2,3,4]

L.reverse()
print(L)
```

## Sort

```python
L = [4,2,1,55,10]
L.sort()
print(L)
```

## Copy

```python
L = [1,2,3,4]

l1 = L.copy()

print(l1)
```
