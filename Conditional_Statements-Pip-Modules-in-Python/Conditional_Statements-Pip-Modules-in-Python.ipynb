## Conditional Statements in Python

```python
### if - elif - else
"""
if condition:
    write something
elif condition:
    write something
else:
    write something
"""
```

```python
# Login Function Demo - riduan.com

"""
Actual Credentials:
Email: riduan.aziz@gmail.com
Password: riduan@123
"""

email = input("Enter your email: ")
password = input("Enter your password: ")

actual_email = "riduan.aziz@gmail.com"
actual_password = "riduan@123"

if email == actual_email and password == actual_password:
    print("Login Successfull!")

elif email == actual_email and password != actual_password:
    print("Incorrect Password!")

else:
    print("Invalid Credentials!")
```

## Nested Conditional Statements in Python

```python
### if - elif - else
"""
if condition:
    write something
elif condition:
    write something
    if codition:
        write something
    else:
        write something
else:
    write something
"""
```

```python
# Login Function Demo - riduan.com

"""
Actual Credentials:
Email: riduan.aziz@gmail.com
Password: riduan@123
"""

email = input("Enter your email: ")
password = input("Enter your password: ")

actual_email = "riduan.aziz@gmail.com"
actual_password = "riduan@123"

if email == actual_email and password == actual_password:
    print("Login Successfull!")

elif email == actual_email and password != actual_password:
    print("Wrong password: Please try again!")
    password = input("Enter your password again: ")
    
    if password == actual_password:
        print("Login Successfull!")
    else:
        print("Again Wrong password!")

else:
    print("Invalid Credentials!")
```
