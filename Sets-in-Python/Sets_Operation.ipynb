# Sets Operations

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
```

## Union(|)

```python
s1 | s2
```

## Intersection(&)

```python
s1 & s2
```

## Difference(-)

```python
s1 - s2
```

## Symmetric Difference(^)

```python
s1 ^ s2
```

# Membership

```python
1 in s1
```

```python
1 not in s1
```

## Iteration

```python
for i in s1:
    print(i)
```
