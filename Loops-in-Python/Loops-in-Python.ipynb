## For Loop

```python
for i in range(1,11):
    print(i)
```

### Forward Loop

```python
for i in range(1,11):
    print(i, "Riduan")
```

### Reverse Loop

```python
for i in range(11,-1,-1):
    print(i)
```

### Jump parameter


```python
for i in range(1,11,2):
    print(i)
```

### Character

```python
for i in "Riduan":
    print(i)
```

### Lists

```python
for i in [1,2,3,4]:
    print(i)
```

```python
for i in (1,2,3,4):
    print(i)
```

## Nested For Loops

```python
for i in range(1,5):
    for j in range(1,5):
        print(i,j)
```

## Exercise: Print Triangle Coordinates

```python
row = int(input("Enter the number of rows: "))

for i in range(1, row+1):
    for j in range(1, i+1):
        print("*", end="")
    print("")
```

## Loop Control Statements

- Break
- Continue
- Pass

```python
# Break

for i in range(1,11):
    if i == 5:
        break
    print(i)
```

```python
name = ["Riduan", "Adury", "Hosna", "Aziz" ]

for i in name:
    if i == "Riduan":
        print(i)
        break
```

```python
# Continue

for i in range(1,10):
    if i == 5:
        continue
    print(i)
```

```python
 # Pass

for i in range(1,11):
    pass
```
