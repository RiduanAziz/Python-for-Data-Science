## Strings are sequence of Characters

#### In Python specifically, strings are a sequence of Unicode Characters

Topic to be discussed:

- Creating Strings
- Accessing Strings

- Editing Strings
- Deleting Strings
- Operations on Strings
- String Functions

## Creating Stings

```python
s = 'hello'
s = "hello"
# multiline strings
s = '''hello'''
s = """hello"""
s = str('hello')
print(s)
```

```python
a = 34
str(a)
```

### Single Line

```python
text = "it's my bike"
print(text)
```

### Multiline Line

```python
text = """it's my bike
i love my bike"""

print(text)
```

## Accessing Strings

```python
s = "welcome to python"
```

### positive access

```python
print(s[13])
```

```python
len(s)
```

### Negative indexing

```python
print(s[-11])
```

### String slicing

```python
s[:]
```

```python
s[11:15]
```

### String Jump

```python
print(s[0:7:2])
```

### Reverse Slicing

```python
print(s[6::-1])
```

```python
# Full Reverse
print(s[::-1])
```

```python
# Full reverse with one character skip
print(s[::-2])
```

```python
# ** Python Strings are Immutable - it means not changeable **
s[0] = "A"
```
