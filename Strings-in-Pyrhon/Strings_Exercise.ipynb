3. Write a program that can check whether a given string is palindrome or not.

 - abba
 - mam
 - madam

```python
word = input("Enter your word: ")

flag = True
for i in range (0, len(word)//2):
    if word[i] != word[len(word) - i -1]:
        flag = False
        print("Not a Palindrome")
        break
if flag:
    print("Palindrome")
```
